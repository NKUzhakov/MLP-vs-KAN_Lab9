## 3. Теоретичні відомості

### Багатошаровий персептрон.
MLP є різновидом штучної нейронної мережі, які мають базову або загальну архітектуру. Основою MLP є окремий нейрон. Ці обчислювальні одиниці приймають вхідні дані, підсумовують їх за допомогою зважених зв’язків і застосовують функцію активації для створення вихідних даних. Цей повторюваний процес на всіх рівнях дозволяє мережі виділяти все більш складні ознаки з даних, з якими вона працює. На відміну від одношарових персептронів, які обмежені лінійними зв’язками, MLP з їхніми прихованими шарами можуть опрацьовувати нелінійні залежності, що є досить важливим аспектом для багатьох реальних задач.

Ця універсальність зробила MLP застосовною у багатьох сферах штучного інтелекту. У розпізнаванні зображень MLP використовуються в системах розпізнавання облич, автономного керування транспортними засобами та аналізу медичних зображень. Їх здатність виявляти тонкі закономірності в значеннях пікселів забезпечує точну ідентифікацію об’єктів і виявлення аномалій. У сфері обробки природної мови MLP застосовуються в механізмах машинного перекладу, аналізу настроїв і взаємодії з чат-ботами. Їх вміння розпізнавати нюанси людської мови сприяє розвитку інтелектуального спілкування та витягування інформації.

Дана архітектура складається з мінімум трьох шарів взаємопов’язаних вузлів, що називаються нейронами. Вхідний шар приймає початкові дані, наприклад, пікселі зображення або слова в реченні. Прихований шар використовується для складних обчислень та вилучення ознак. Таких шарів може бути один або кілька, кожен з яких має різну кількість нейронів. Останнім є вихідний шар, що видає остаточний прогноз або результат, заснований на обробці, виконаній на попередніх шарах. Кожен нейрон використовує функцію активації для своєї зваженої суми входів. Ця функція додає нелінійність, що дозволяє мережі навчатися складним взаємозв’язкам у даних. Популярні функції активації включають сигмоїдну і ReLU.
 
MLP навчається через процес зворотнього поширення. Під час навчання мережа робить прогнози на основі набору даних, порівнює їх з фактичними значеннями, а потім коригує ваги в зв’язках, щоб мінімізувати помилку. Цей ітеративний процес поступово налаштовує мережу для точного зіставлення входів з бажаними виходами.

### Нейрона мережа Колмогорова-Арнольда 
KAN – це нейронні мережі, що базуються на теоремі Колмогорова-Арнольда про представлення функцій багатьх змінних. Ця теорема стверджує, що кожна багатовимірна неперервна функція може бути представлена сумою неперервних одновимірних функцій. Вона говорить, що кожну складну багатовимірну функцію можна розкласти на простіші одновимірні функції. 

У своїй найпростішій формі KAN подібний до рівняння теореми Колмогорова-Арнольда і складається лише з двох шарів. Перший рівень перетворює кожен вхідний сигнал за допомогою набору одновимірних функцій. Другий рівень підсумовує ці перетворення і виводить остаточний прогноз. Але іноді для вивчення складних функцій реального світу, KAN складаються з декількох шарів, так само як і MLP, де вихід кожного шару є входом для наступного шару. Багатошарові KAN складаються з: вхідного шару, ребер, де виконується більшість обчислень, та вузлів. На відміну від MLP, де кожне ребро має асоційований ваговий параметр, в KAN ці ваги повністю замінюються функціями, що навчаються, – одновимірними функціями.
Ці одновимірні функції параметризуються за допомогою B-сплайна.

B-сплайни - це математичні функції, які дозволяють з’єднати набір точок за допомогою плавної кривої. Bони складаються з точок, які називаються контрольними точками, що допомагають гнучко налаштовувати локальні сегменти цієї кривої, не впливаючи на загальну криву.
 
Отже, ці одновимірні функції можна розглядати як функції активації, що навчаються, присутні на ребрах нейронної мережі. Вузли в KAN просто виконують підсумовування вхідних сигналів замість того, щоб застосовувати будь-яку функцію активації. Оскільки всі операції, що відбуваються в KAN, є диференційованими, їх можна навчати за допомогою зворотного поширення зі звичайними функціями втрат.

## 4. Виконання роботи

### 4.1. Створення MLP 

Спочатку створимо модель нейронної мережі прямого поширення, для чого визначимо відповідний клас. При ініціалізації можна задати її архітектуру, передавши список кількостей нейронів відповідних рівнів. Також у якості функції активації за замовчуванням використовується ReLU. Функція активації є однаковою на кожному рівні мережі.

```python
import torch
from torch import Tensor
from torch.nn import Linear, Module, ModuleList, ReLU

class MLP(Module):
    def __init__(self, layers, act_fn = ReLU):
        super().__init__()
        self.act_fn = act_fn()
        self.layers = ModuleList()
        for in_features, out_features in zip(layers, layers[1:]):
            self.layers.append(Linear(in_features, out_features))

    def forward(self, x: Tensor):
        for layer in self.layers:
            x = self.act_fn(layer(x))
        return x
```

### 4.2. Створення KAN

Наступним кроком створимо мережу Колмогорова-Арнольда, де замість простого множення на ваги використовуються В-сплайни. Але для цього спочатку реалізуємо відповідний шар даної мережі.

```python
from math import sqrt
from torch import (arange, no_grad, rand, linalg, bmm, sort, linspace, concatenate, float32, int64,
                   sum as torch_sum)
from torch.nn import SiLU, Parameter, init, functional, ModuleList

class KANLayer(Module):
    def __init__(self, in_features, out_features, grid_size=5, spline_order=3, scale_noise=0.1, scale_base=1.0,
                 scale_spline=1.0, enable_standalone_scale_spline=True, base_activation=SiLU, grid_eps=0.02,
                 grid_range=(-1, 1)):
        super().__init__()
        self.in_features = in_features
        self.out_features = out_features

        self.grid_size = grid_size
        self.spline_order = spline_order
        self.scale_noise = scale_noise        
        self.scale_spline = scale_spline
        self.enable_standalone_scale_spline = enable_standalone_scale_spline        
        self.grid_eps = grid_eps

        self.scale_base = scale_base
        self.base_activation = base_activation()

        grid_step = (grid_range[1] - grid_range[0]) / grid_size
        grid = ((arange(-spline_order, grid_size + spline_order + 1) * grid_step + grid_range[0]).expand(in_features,-1).contiguous())
        self.register_buffer("grid", grid) # registering it as a model's state

        self.base_weight = Parameter(Tensor(out_features, in_features)) # weights for activation part
        self.spline_weight = Parameter(Tensor(out_features, in_features, grid_size + spline_order)) # coefficient of the splines
        if enable_standalone_scale_spline:
            self.spline_scaler = Parameter(Tensor(out_features, in_features))

        self.reset_parameters() # need to init the params as they have the random values from RAM

    def reset_parameters(self):
        init.kaiming_uniform_(self.base_weight, a=sqrt(5) * self.scale_base)
        with no_grad():
            noise = ((rand(self.grid_size + 1, self.in_features, self.out_features) - 1 / 2) * self.scale_noise / self.grid_size)
            self.spline_weight.data.copy_((self.scale_spline if not self.enable_standalone_scale_spline else 1.0) 
                                          # need to build the splines according to the new coefs
                                          * self.curve2coeff(self.grid.T[self.spline_order: -self.spline_order], noise)) 
            if self.enable_standalone_scale_spline:
                init.kaiming_uniform_(self.spline_scaler, a=sqrt(5) * self.scale_spline)

    def curve2coeff(self, X: Tensor, Y: Tensor):        
        # Finds the coefficients of the curve that interpolates the given points.        
        assert X.dim() == 2 and X.size(1) == self.in_features
        assert Y.size() == (X.size(0), self.in_features, self.out_features)

        A = self.b_splines(X).transpose(0, 1)  
        B = Y.transpose(0, 1)
        result = linalg.lstsq(A, B).solution.permute(2, 0, 1) 

        assert result.size() == (self.out_features, self.in_features, self.grid_size + self.spline_order,)
        return result.contiguous()
    
    def b_splines(self, x: Tensor):
        # Finds the B-spline bases for the given input tensor.
        assert x.dim() == 2 and x.size(1) == self.in_features

        grid: Tensor = self.grid  
        x = x.unsqueeze(-1)
        bases = ((x >= grid[:, :-1]) & (x < grid[:, 1:])).to(x.dtype) # divided difference of 0 order
        for k in range(1, self.spline_order + 1):
            # recurrently calc the divided differences of higher orders
            bases = ((x - grid[:, : -(k + 1)]) / (grid[:, k:-1] - grid[:, : -(k + 1)]) * bases[:, :, :-1]) + (
                    (grid[:, k + 1:] - x) / (grid[:, k + 1:] - grid[:, 1:(-k)]) * bases[:, :, 1:])

        assert bases.size() == (x.size(0), self.in_features, self.grid_size + self.spline_order,)
        return bases.contiguous()

    @property
    def scaled_spline_weight(self):
        return self.spline_weight * (self.spline_scaler.unsqueeze(-1) if self.enable_standalone_scale_spline else 1.0)

    def forward(self, x: Tensor):
        assert x.dim() == 2 and x.size(1) == self.in_features

        base_output = functional.linear(self.base_activation(x), self.base_weight)
        spline_output = functional.linear(self.b_splines(x).view(x.size(0), -1), self.scaled_spline_weight.view(self.out_features, -1))
        return base_output + spline_output
```

Наступним кроком за аналогією до мережі прямого поширення визначимо модель KAN

```python
class KAN(Module):
    def __init__(self, layers, grid_size=5, spline_order=3, scale_noise=0.1, scale_base=1.0, scale_spline=1.0,
                 base_activation=SiLU, grid_eps=0.02, grid_range=(-1, 1)):
        super().__init__()
        self.grid_size = grid_size
        self.spline_order = spline_order

        self.layers = ModuleList()
        for in_features, out_features in zip(layers, layers[1:]):
            self.layers.append(KANLayer(in_features, out_features, grid_size=grid_size, spline_order=spline_order,
                                         scale_noise=scale_noise, scale_base=scale_base, scale_spline=scale_spline,
                                         base_activation=base_activation, grid_eps=grid_eps, grid_range=grid_range, ))

    def forward(self, x: Tensor):
        for layer in self.layers:
            x = layer(x)
        return x
```

### 4.3 Довизначення архітектури моделей та їх створення

Наступним кроком визначимо кількість шарів для обох моделей: вхідний шар, п'ять прихованих та один вихідний шар. Вхідний шар має стільки входів, скільки пікселів мають зображення з набору даних, у прихованих шарах кількість нейронів становлено 20, виходів є рівно стільки скільки класів у наборі даних (рукописні цифри від 0 до 9).

```python
input_size = 28 * 28
num_classes = 10

hidden_layers_num = 5
hidden_size = 20
nn_layers = [input_size] + [hidden_size for _ in range(hidden_layers_num)] + [num_classes]
```

Далі створимо екземпляри визначених моделей із налаштуванням вищезазначених параметрів.

```python
MLP_model = MLP(nn_layers)
KAN_model = KAN(nn_layers)
```

### 4.4 Навчання створених моделей

Визначимо параметри навчання обох моделей. Швидкість навчання встановлено 0.001, достатньо мала що зменшує ризик "перескочити" мінімум функції втрат та дає більш плавне та стабільне навчання. Порція даних (batch_size) - кількість зразків, що обробляється одночасно перед оновленням ваги моделі. Обрано 32 – не занадто мале для пришвидшення навчання, але не надто велика для забезпечення більшої точності. Кількість епох (разів проходжень по даним) обрано 50, що є достатнім щоб модель встигла навчитись. Також параметр weight_decay використовується для регуляризації: він забезпечує регулярне але невелике зменшення вагів моделі для запобігання перенавчання. У якості функції втрат можна використати міжкласову ентропію, що визначає міру розбіжності між прогнозованим та дійсним класом.

```python
from torch.nn import CrossEntropyLoss
learning_rate = 0.001
batch_size = 32
epochs = 50
weight_decay = 1e-4

loss_func = CrossEntropyLoss()
```

Після цього оберемо та налаштуємо оптимізатор моделей а також правило, за яким буде поступово зменшуватись швидкість навчання.

```python
from torch.optim import AdamW, Optimizer
from torch.optim.lr_scheduler import ExponentialLR

MLP_optimizer = AdamW(MLP_model.parameters(), lr=learning_rate, weight_decay=weight_decay)
MLP_scheduler = ExponentialLR(MLP_optimizer, gamma=0.8)

KAN_optimizer = AdamW(KAN_model.parameters(), lr=learning_rate, weight_decay=weight_decay)
KAN_scheduler = ExponentialLR(KAN_optimizer, gamma=0.8)
```

Далі завантажимо тестові дані з датасету MNIST, що містить близько 60000 чорно-білих зображень рукописних цифр у розмірі 28х28. Після чого створимо об'єкти обгортки для отримання даних за поціями. Слід зазначити, що дані нормуються при конвертації в тензор, тобто значення пікселів знаходиться в межах [0, 1].

```python
from torch import cuda
from torch.utils.data import DataLoader, random_split
from torchvision.datasets import MNIST
from torchvision.transforms import ToTensor

train_dataset = MNIST(root="./data", train=True, download=True, transform=ToTensor())
test_dataset = MNIST(root="./data", train=False, transform=ToTensor())
val_dataset, test_dataset = random_split(test_dataset, [1_000, 9_000])

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=True)

device = torch.device("cuda" if cuda.is_available() else "cpu")
```

Визначимо функцію для навчання моделей з урахуванням вищезазначених параметрів

```python
def train(model, optimizer, scheduler, train_loader, val_loader):   
    train_loss_all = [] 
    val_loss_all = [] 
    for epoch in range(epochs): 
        train_loss = 0 
        train_num = 0 
        
        model.train() 
         
        for images, labels in train_loader: 
            images = images.reshape(-1, input_size).to(device) 
            labels = labels.to(device) 

            optimizer.zero_grad() 
            outputs = model(images) 
            
            loss = loss_func(outputs, labels) 
            loss.backward() 
            optimizer.step() 
            
            train_loss += loss.item() * images.size(0) 
            train_num += images.size(0) 
        
        train_loss_all.append(train_loss / train_num) # average loss of epoch   
        
        model.eval() 
        val_loss = 0 
        val_num = 0 
        with torch.no_grad(): 
            for images, labels in val_loader: 
                images = images.reshape(-1, input_size).to(device) 
                labels = labels.to(device) 
                outputs = model(images) 

                val_loss += loss_func(outputs, labels).item() * images.size(0) 
                val_num += images.size(0) 
        val_loss_all.append(val_loss / val_num) 

        scheduler.step() # next epoch
    return (train_loss_all, val_loss_all)
```

Запустимо торенування та збережемо отримані моделі.

```python
import time

MLP_start = time.perf_counter()
MLP_train_loss_all , MLP_val_loss_all = train(MLP_model, MLP_optimizer, MLP_scheduler, train_loader, val_loader)
MLP_time = time.perf_counter() - MLP_start
print(f"MLP tarining time: {MLP_time}s")

KAN_start = time.perf_counter()
KAN_train_loss_all, KAN_val_loss_all = train(KAN_model, KAN_optimizer, KAN_scheduler, train_loader, val_loader)
KAN_time = time.perf_counter() - KAN_start
print(f"KAN tarining time: {KAN_time}s")

torch.save(MLP_model.state_dict(), "./models/mlp_mnist.pth")
torch.save(KAN_model.state_dict(), "./models/kan_mnist.pth")
```

Отже можна побачити, що навчання мережі KAN зайняло значно більше часу порівняно з мережею MLP. Це пов'язано із тим, що у даних моделей є однаковою кількість вагів у MLP та В-сплайнів у KAN, і друга має більше параметрів.

Наведемо графіки значень функції втрат на тренувальних та валідаційних даних за епохами

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(15, 4))
axes[0].plot(MLP_train_loss_all, label="Train Loss")
axes[0].plot(MLP_val_loss_all, label="Val Loss")
axes[0].set_title("MLP")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].legend()

axes[1].plot(KAN_train_loss_all, label="Train Loss")
axes[1].plot(KAN_val_loss_all, label="Val Loss")
axes[1].set_title("KAN")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].legend()

plt.show()
```

Можна помітити, що похибка на тренувльних даних у обох моделей менша за відповідні похибки на валідаційному наборі. Це пояснюється тим, що другий набір моделі не 'бачать'. Далі, перевіримо точність класифікації обох моделей, використовуючи метрики Accuracy, Precision, Recall та F1:
* Accuracy означає відсоток об'єктів, що класифіковано правильно; 
* Precision - відсоток правильних класів серед передбачених об'єктів; 
* Recall - відсоток віднайдених серед усіх реальних об'єктів класів; 
* F1 - середня міра між precision і recall.

```python
from torchmetrics.classification import MulticlassAccuracy, MulticlassPrecision, MulticlassRecall, MulticlassF1Score

def classification_report(preds, labels, model_name):
    acc = MulticlassAccuracy(num_classes=num_classes)
    precision = MulticlassPrecision(num_classes=num_classes, average='macro')
    recall = MulticlassRecall(num_classes=num_classes, average='macro')
    f1 = MulticlassF1Score(num_classes=num_classes, average='macro')

    accuracy_val = acc(preds, labels)
    precision_val = precision(preds, labels)
    recall_val = recall(preds, labels)
    f1_val = f1(preds, labels)

    print(f"{model_name}:")
    print("\tAccuracy:\t", accuracy_val.item())
    print("\tPrecision:\t", precision_val.item())
    print("\tRecall:  \t", recall_val.item())
    print("\tF1 Score:\t", f1_val.item())
```

### 4.6 KAN: перевірка ефективності B-сплайнів та поліномів Лагранжа

Спочатку трохи модифікуємо шар мережі KAN, де замість В-сплайнів будуть використовуватися інтерполяційні поліноми у формі Лагранжа.

```python
class KANLayer_Lagrange(Module):
    def __init__(self, in_features, out_features, grid_size=5, scale_noise=0.1, scale_base=1.0,
                 scale_polynom=1.0, enable_standalone_scale_polynom=True, base_activation=SiLU, grid_eps=0.02,
                 grid_range=(-1, 1)):
        super().__init__()
        self.in_features = in_features
        self.out_features = out_features

        self.grid_size = grid_size
        self.scale_noise = scale_noise        
        self.scale_polynom = scale_polynom
        self.enable_standalone_scale_spline = enable_standalone_scale_polynom        
        self.grid_eps = grid_eps

        self.scale_base = scale_base
        self.base_activation = base_activation()

        grid = torch.linspace(grid_range[0], grid_range[1], grid_size).expand(in_features, -1).contiguous()

        self.register_buffer("grid", grid) # registering it as a model's state

        self.base_weight = Parameter(Tensor(out_features, in_features)) # weights for activation part
        self.polynom_weight = Parameter(Tensor(out_features, in_features, grid_size)) # y_i
        if enable_standalone_scale_polynom:
            self.polynom_scaler = Parameter(Tensor(out_features, in_features))

        self.reset_parameters() # need to init the params as they have the random values from RAM

    def reset_parameters(self):
        with no_grad():
            init.kaiming_uniform_(self.base_weight, a=sqrt(5) * self.scale_base)
            init.kaiming_uniform_(self.polynom_weight, a=0.00000001 * self.scale_polynom)

            if self.enable_standalone_scale_spline:
                init.kaiming_uniform_(self.polynom_scaler, a=0.00000001 * self.scale_polynom)
    
    def lagrange_bases(self, x: Tensor):
        # Finds the B-spline bases for the given input tensor.
        assert x.dim() == 2 and x.size(1) == self.in_features

        grid: Tensor = self.grid  
        bases = torch.ones(x.size(0), self.in_features, self.grid_size, device=x.device)
        
        for i in range(self.grid_size):
            for j in range(self.grid_size):
                if i != j:
                    bases[:, :, i] *= (x - grid[:, j]) / (grid[:, i] - grid[:, j] + 0.01)
        return bases.contiguous()

    @property
    def scaled_polynom_weight(self):
        return self.polynom_weight * (self.polynom_scaler.unsqueeze(-1) if self.enable_standalone_scale_spline else 1.0)

    def forward(self, x: Tensor):
        assert x.dim() == 2 and x.size(1) == self.in_features
        assert not torch.isnan(x).any()
        x = torch.clamp(x, self.grid[:, 0], self.grid[:, -1])

        base_output = functional.linear(self.base_activation(x), self.base_weight)
        polynom_output = functional.linear(self.lagrange_bases(x).view(x.size(0), -1), self.scaled_polynom_weight.view(self.out_features, -1))

        return base_output + polynom_output
```

Далі, за аналогією до моделі KAN створимо модель, яка викоримтовуватиме поліноми Лагранжа (окремий клас було зроблено через інший набір параметрів у конструкторі).

```python
class KAN_Lagrange(Module):
    def __init__(self, layers, grid_size=5, scale_noise=0.1, scale_base=1.0, scale_polynom=1.0,
                 base_activation=SiLU, grid_eps=0.02, grid_range=(-1, 1)):
        super().__init__()
        self.grid_size = grid_size

        self.layers = ModuleList()
        for in_features, out_features in zip(layers, layers[1:]):
            self.layers.append(KANLayer_Lagrange(in_features, out_features, grid_size=grid_size,
                                         scale_noise=scale_noise, scale_base=scale_base, scale_polynom=scale_polynom,
                                         base_activation=base_activation, grid_eps=grid_eps, grid_range=grid_range, ))

    def forward(self, x: Tensor):
        for layer in self.layers:
            x = layer(x)
        return x
```

Створимо дану молель із раніше зазначеною кількістю шарів та нейронів на кожному шарі. Аналогічним чином налаштуємо оптимізатор моделей та правило, за яким буде поступово зменшуватись швидкість навчання. Запустимо тренування та збережемо отриману модель.

```python
KAN_Lagrange_model = KAN_Lagrange(nn_layers)

KAN_Lagrange_optimizer = AdamW(KAN_Lagrange_model.parameters(), lr=learning_rate, weight_decay=weight_decay)
KAN_Lagrange_scheduler = ExponentialLR(KAN_Lagrange_optimizer, gamma=0.8)

KAN_Lagrange_start = time.perf_counter()
KAN_Lagrange_train_loss_all, KAN_Lagrange_val_loss_all = train(KAN_Lagrange_model, KAN_Lagrange_optimizer, KAN_Lagrange_scheduler, train_loader, val_loader)
KAN_Lagrange_time = time.perf_counter() - KAN_Lagrange_start
print(f"KAN_Lagrange tarining time: {KAN_Lagrange_time}s")

torch.save(KAN_Lagrange_model.state_dict(), "./models/kan_lagrange_mnist.pth")
```

Також у моделі KAN, що використовує В-сплайни, обидва показники менші, що свідчить про вищу точність такої моделі.
